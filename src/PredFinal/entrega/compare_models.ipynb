# Comparación modelos

> Uso de test de Wilcoxon para comparar modelos

```r
# Ganancias maximas obtenidas por semilla (10) en cada uno de los experimentos
estandarizar_bug     <- c(90965, 92627, 91381, 93042, 88889, 93541, 89470, 88473, 89802, 91879)
deflacion_baseline   <- c(80830, 86230, 87310, 84818, 90716, 83738, 81412, 92128, 87310, 91796)
dolar_blue           <- c(86978, 83821, 83406, 84070, 83073, 84569, 89387, 87559, 82243, 83406)
estandarizar_sin_bug <- c(87809, 87559, 87809, 90716, 90633, 91298, 90550, 89055, 86147, 83904)
ninguno              <- c(87975, 85566, 88390, 87975, 85399, 91796, 83987, 81744, 86812, 83406)

vectors <- list(
  estandarizar_bug        = estandarizar_bug,
  deflacion_baseline      = deflacion_baseline,
  dolar_blue              = dolar_blue,
  estandarizar_sin_bug    = estandarizar_sin_bug,
  ninguno                 = ninguno
)

# Get all pairwise combinations (no repeats, no self-pairs)
combos <- combn(names(vectors), 2, simplify = FALSE)

# Run wilcox test for each combination and store results
results <- lapply(combos, function(pair) {
  res <- wilcox.test(
    vectors[[pair[1]]],
    vectors[[pair[2]]],
    paired = TRUE
  )
  data.frame(
    A         = pair[1],
    B         = pair[2],
    #statistic = res$statistic,
    p_value   = res$p.value
  )
})

# Combine into a single data frame
results_df <- do.call(rbind, results)
rownames(results_df) <- NULL
print(results_df)
```
